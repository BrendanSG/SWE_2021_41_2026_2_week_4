### Happy Number
Write an algorithm to determine if a number **n** is happy.

A **happy number** is a number defined by the following process:

* Starting with any positive integer, replace the number by the sum of the squares of its digits.
* Repeat the process until the number equals 1 (where it will stay), or it **loops endlessly in a cycle** which does not include 1.
* Those numbers for which this process **ends in 1** are happy.

Return **true** if n is a happy number, and **false** if not.

Constraints:
1 <= n <= $2^{31}$ - 1

```python
def isHappy(n):
  # print(f"isHappy({n}) ?")
  # Ensure that n respect constraints
  if n < 1 or n > 2**31-1:
    # print(f"{n} doesn't respect constraints (1 <= n <= 2^31-1)")
    return False

  numbers_visited = set()
  # print(n in numbers_visited)
  while not (n in numbers_visited):
    # print(f"- n={n}:")
    numbers_visited.add(n) # mark n as visited

    # Compute the sum of squares digits
    oldN = n
    n = 0
    while oldN > 0:
      n += (oldN%10)**2
      oldN = oldN//10

    # If we hit 1, break (else statement isn't triggered)
    if n == 1:
      break
  else:
    # if we reach here, it means that we visited the same number twice
    # => loop detected
    # print(f"Loop detected on n={n} !")
    # print(numbers_visited.keys())
    return False

  # print("Process ends on n=1 !")
  # print(numbers_visited.keys())
  return True

# n=19, expected: True, visited numbers: 19 -> 82 -> 68 -> 100 -> 1 (hit)
print(f"isHappy(19) = {isHappy(19)} \t[EXPECTED: TRUE]")
# n=2, expected: False, visited numbers: 2 -> 2 (loop on 2)
print(f"isHappy(2)  = {isHappy(2)} \t[EXPECTED: FALSE]")
# n=22, expected: False, visited numbers: 22 -> 8 -> 64 -> 52 -> 29 -> 85 -> 89 -> 145 -> 42 -> 20 -> 4 -> 16 -> 37 -> 58 -> 89 (loop on 89)
print(f"isHappy(22) = {isHappy(22)} \t[EXPECTED: FALSE]")
# n=18, expected: False, visited numbers: 18 -> 65 -> 61 -> 37 -> 58 -> 89 -> 145 -> 42 -> 20 -> 4 -> 16 -> 37 (loop on 37)
print(f"isHappy(18) = {isHappy(18)} \t[EXPECTED: FALSE]")
# n=21, expected: False, visited numbers: 21 -> 5 -> 25 -> 29 -> 85 -> 89 -> 145 -> 42 -> 20 -> 4 -> 16 -> 37 -> 58 -> 89 (loop on 89)
print(f"isHappy(21) = {isHappy(21)} \t[EXPECTED: FALSE]")
# n=49, expected: True, visited numbers: 49 -> 97 -> 130 -> 10 -> 1 (hit)
print(f"isHappy(49) = {isHappy(49)} \t[EXPECTED: TRUE]")
```
